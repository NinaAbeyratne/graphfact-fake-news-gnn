```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import kagglehub
import re

print("Setup completed successfully")
```

## Load Cleaned Dataset

```python
dt = pd.read_csv('/content/cleaned.csv')
dt.head(3)
```

## Remove URLs

```python
def remove_urls(text):
    return re.sub(r'https?://\S+|www\.\S+', '', text)
```

## Remove HTML Tags

```python
def remove_html(text):
    return re.sub(r'<[^>]+>', '', text)
```

## Normalize Spaces

```python
def normalize_spaces(text):
    return re.sub(r'\s+', ' ', text).strip()
```

## clean_text() pipeline

```python
def clean_text(text):

    text = text.lower()

    # remove URLs
    text = remove_urls(text)

    # remove HTML tags
    text = remove_html(text)

    # normalize whitespace
    text = normalize_spaces(text)

    return text

# apply to both columns
dt['statement'] = dt['statement'].apply(clean_text)
dt['context'] = dt['context'].apply(clean_text)
```

```python
dt["article"] = (
    dt["context"].fillna("") + " " + dt["statement"].fillna("")
).str.strip()
```

```python
def truncate_words(text, max_words=128):
    if not isinstance(text, str):
        return text
    return " ".join(text.split()[:max_words])

dt["article"] = dt["article"].apply(truncate_words)
```

```python
dt = dt.drop(columns=['statement', 'context'])
dt.sample(3)
```
